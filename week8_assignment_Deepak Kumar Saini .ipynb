# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus
- Improve routing
- Add logging
- Add more tools


````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
# 🤖 AGENT FUNCTION

def agent(query: str):
    query_lower = query.lower()

    try:
        # Route to Calculator
        if "calculate" in query_lower:
            expression = query_lower.replace("calculate", "").strip()

            return {
                "type": "calculation",
                "result": calculator(expression)
            }

        # Route to Keyword Extractor
        elif "keywords" in query_lower:
            text = query.replace("Extract keywords from", "")
            text = text.replace("extract keywords from", "").strip()

            return {
                "type": "keywords",
                "result": extract_keywords(text)
            }

        # General Query
        else:
            return {
                "type": "general",
                "result": "This is a general query. No tool required."
            }

    except Exception as e:
        return {
            "type": "error",
            "result": str(e)
        }
````

## 📦 Expected Output Format

```
{
  "type": "calculation / keywords / general / error",
  "result": ...
}
```

````python
# 🧪 Test Cases

queries = [
    "Calculate 20 + 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "What is machine learning?"
]

for q in queries:
    print("Query:", q)
    print("Response:", agent(q))
    print("-" * 50)
````

````python
# 🎯 Interactive Mode

while True:
    user_input = input("Enter query (type 'exit' to stop): ")

    if user_input.lower() == "exit":
        print("Goodbye!")
        break

    print("Response:", agent(user_input))
````

# Bonus

````python
# 🛠️ BONUS TOOL: Word Counter

def word_count(text: str) -> int:
    """Count the number of words in a sentence."""
    try:
        return len(text.split())
    except Exception:
        return 0
````

````python
# 🤖 IMPROVED AGENT WITH BONUS FEATURES

def agent(query: str):
    query_lower = query.lower()

    print(f"\n[INFO] Received Query: {query}")

    try:

        # Calculator Routing
        if any(word in query_lower for word in
               ["calculate", "add", "subtract", "multiply", "divide", "+", "-", "*", "/"]):

            print("[INFO] Routing -> Calculator Tool")

            if "calculate" in query_lower:
                expression = query_lower.replace("calculate", "").strip()
            else:
                expression = query

            result = calculator(expression)

            return {
                "type": "calculation",
                "result": result
            }

        # Keyword Extraction Routing
        elif "keywords" in query_lower:

            print("[INFO] Routing -> Keyword Extractor Tool")

            text = query.replace("Extract keywords from", "")
            text = text.replace("extract keywords from", "").strip()

            result = extract_keywords(text)

            return {
                "type": "keywords",
                "result": result
            }

        # Word Count Tool
        elif "count words" in query_lower:

            print("[INFO] Routing -> Word Counter Tool")

            text = query.replace("count words", "").strip()

            result = word_count(text)

            return {
                "type": "word_count",
                "result": result
            }

        # General Query
        else:

            print("[INFO] Routing -> General Response")

            return {
                "type": "general",
                "result": "This is a general query. No tool was required."
            }

    except Exception as e:

        print("[ERROR]", e)

        return {
            "type": "error",
            "result": str(e)
        }
````

````python
queries = [
    "Calculate 25 * 4",
    "Extract keywords from Artificial Intelligence is changing healthcare and education",
    "Count words ChatGPT makes learning easier",
    "What is Deep Learning?"
]

for q in queries:
    print("=" * 60)
    print("Query:", q)
    print("Response:", agent(q))
````

````python
while True:

    user_input = input("\nEnter query (type 'exit' to stop): ")

    if user_input.lower() == "exit":
        print("Goodbye!")
        break

    print(agent(user_input))
````
